```python
!pip install chromadb sentence-transformers

import chromadb
from sentence_transformers import SentenceTransformer

# Creamos la base de datos vectorial
chroma_client = chromadb.Client()
collection = chroma_client.get_or_create_collection(name="Tecsup")

# Motor de embeddings inicial
embedder = SentenceTransformer('all-MiniLM-L6-v2')

# Información de los compañeros
informacion = [
    "Me llamo Fiorella Yanina Trujillo Tello, estudio Big Data en Tecsup, vivo en Miraflores y me gusta la fotografía y los viajes.",
    "Soy César Diego Ruelas Flores, estudiante de Big Data en Tecsup, vivo en San Borja y me encanta el fútbol y la tecnología.",
    "Mi nombre es Juan Sebastian Rosales Castro, estudio Big Data en Tecsup, vivo en Surco y me apasiona la música y tocar guitarra.",
    "Soy Juan Antonio Bustamante Guerra, estudio Big Data en Tecsup, vivo en San Juan de Lurigancho y me gusta el ciclismo y el cine.",
    "Me llamo Joshua Jian Cora Salazar, estudiante de Big Data en Tecsup, vivo en Pueblo Libre y me gustan los videojuegos y la robótica.",
    "Soy Bryan Jesus Hinostroza Martinez, estudiante de Big Data en Tecsup, vivo en Los Olivos y me gusta el skate y la programación.",
    "Me llamo Abraham Jhonatan Huamán Quispe, estudio Big Data en Tecsup, vivo en Ate y me gusta la cocina y el ajedrez.",
    "Soy Kevin Estiben Olortegui Perez, estudio Big Data en Tecsup, vivo en Magdalena y me gusta el diseño gráfico y los videojuegos.",
    "Mi nombre es Ángela Belén Gutierrez García, estudio Big Data en Tecsup, vivo en Jesús María y me gusta leer novelas y pintar.",
    "Soy Hernán Jhonatan Garro Manayay, estudiante de Big Data en Tecsup, vivo en Villa El Salvador y me gusta el fútbol y la astronomía.",
    "Me llamo Christian David Huaman Escudero, estudio Big Data en Tecsup, vivo en San Miguel y me gusta tocar batería y el cine.",
    "Soy Jesús Zósimo Esplana Sulla, estudio Big Data en Tecsup, vivo en Breña y me gusta la fotografía y correr.",
    "Mi nombre es Sebastian David Travezaño Sánchez, estudio Big Data en Tecsup, vivo en Lince y me gustan los videojuegos y el anime.",
    "Soy Joseph Anccasi Pozo, estudio Big Data en Tecsup, vivo en San Juan de Miraflores y me gusta la programación y el ciclismo.",
    "Me llamo Diego Fuentes Arellano, estudiante de Big Data en Tecsup, vivo en Chorrillos y me interesa la inteligencia artificial y la natación.",
    "Soy Allinson Valeria Varillas Torres, estudio Big Data en Tecsup, vivo en San Isidro y me gustan la moda y la música.",
    "Mi nombre es Gerberth Cuellar Forton, estudio Big Data en Tecsup, vivo en La Molina y me gustan el fútbol y la programación.",
    "Soy Matias Edwin Moreno Taboada, estudio Big Data en Tecsup, vivo en Callao y me gustan el basketball y los videojuegos."
]

# Convertimos a vectores
vectores = embedder.encode(informacion)

# Almacenamos en la base de datos
for i, texto in enumerate(informacion):
    collection.add(
        documents=[texto],
        embeddings=[vectores[i].tolist()],
        ids=[f"doc{i}"]
    )

print("✅ Información guardada en la base de datos 'Tecsup'.")

# Prompt de búsqueda
prompt = "¿Quiénes viven en San Borja o Miraflores?"

# Convertimos el prompt a vector
vector_prompt = embedder.encode([prompt])

# Búsqueda en la base de datos
resultados = collection.query(
    query_embeddings=vector_prompt.tolist(),
    n_results=5
)

print("\n🔎 Resultados encontrados:")
for doc in resultados['documents'][0]:
    print("-", doc)
```

```python
# ---------------------------
# Segundo motor de embeddings
# ---------------------------

print("\n🔧 Probando con otro motor de embedding: 'paraphrase-MiniLM-L3-v2'")

embedder2 = SentenceTransformer('paraphrase-MiniLM-L3-v2')
collection2 = chroma_client.get_or_create_collection(name="Tecsup2")

# Nuevos vectores
vectores2 = embedder2.encode(informacion)

# Guardamos en la nueva colección
for i, texto in enumerate(informacion):
    collection2.add(
        documents=[texto],
        embeddings=[vectores2[i].tolist()],
        ids=[f"doc2_{i}"]
    )

# Búsqueda con el segundo motor
vector_prompt2 = embedder2.encode([prompt])
resultados2 = collection2.query(
    query_embeddings=vector_prompt2.tolist(),
    n_results=5
)

print("\n🔎 Resultados con el segundo motor de embeddings:")
for doc in resultados2['documents'][0]:
    print("-", doc)
```

Al realizar un prompt de búsqueda relacionado a los distritos de residencia, observé que ambos modelos identificaron resultados relevantes, aunque con ligeras variaciones en el orden y selección de las respuestas. Esto demuestra que, si bien ambos motores son eficientes en el proceso de comprensión semántica, existen diferencias en la forma como interpretan y priorizan la información. Finalmente, pude concluir que el motor paraphrase-MiniLM-L3-v2 mostró resultados algo más alineados al sentido general del prompt, mientras que el primero recuperó información útil, pero con una selección un poco más dispersa.
